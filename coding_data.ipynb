```python
import pandas as pd
import os

# ===============================
# 1. File paths
# ===============================
input_file = "1-CSR_CMT_Raw_Data.xlsx"   # keep exact filename
output_file = "CSR_Coded_Data.xlsx"

# ===============================
# 2. Load file
# ===============================
df = pd.read_excel(input_file)

# Clean column names
df.columns = df.columns.str.strip()

# ===============================
# 3. Coding maps for demographics
# ===============================
coding_maps = {
    "AGE": {
        "18-24": 0,
        "25-34": 1,
        "Above 35": 2,
        "18–24": 0,
        "25–34": 1
    },
    "GNDR": {
        "Male": 0,
        "Female": 1
    },
    "SIZE": {
        "Less than 50": 0,
        "50-100": 1,
        "101-200": 2,
        "201-1000": 3,
        "More than 1000": 4,
        "Less than 50 employees": 0
    },
    "REGION": {
        "EURO": 1,
        "Europe": 1,
        "MENA": 2
    },
    "DEGREE": {
        "Undergraduate": 0,
        "Graduate": 1
    }
}

# ===============================
# 4. Likert mapping
# ===============================
likert_map = {
    "Strongly Disagree": 1,
    "Disagree": 2,
    "Neutral": 3,
    "Agree": 4,
    "Strongly Agree": 5,
    "strongly disagree": 1,
    "disagree": 2,
    "neutral": 3,
    "agree": 4,
    "strongly agree": 5
}

# ===============================
# 5. Apply demographic coding
# ===============================
for col, mapping in coding_maps.items():
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().replace(mapping)

# ===============================
# 6. Convert survey columns to numeric
# ===============================
# Adjust this if needed after checking your real columns
survey_prefixes = ["CSRSTK", "CSRCUS", "CSREMP", "EMPID", "CSRIMP", "CMTAFC"]

survey_cols = [
    col for col in df.columns
    if any(col.upper().startswith(prefix) for prefix in survey_prefixes)
]

for col in survey_cols:
    # First clean strings
    df[col] = df[col].astype(str).str.strip()

    # Map Likert text to numeric where applicable
    df[col] = df[col].replace(likert_map)

    # Convert final result to numeric
    df[col] = pd.to_numeric(df[col], errors="coerce")

# ===============================
# 7. Reverse negative questions
# ===============================
# Replace these with the actual negative items from your questionnaire
negative_items = ["CSRSTK2", "CSREMP4"]

for col in negative_items:
    if col in df.columns:
        df[col] = df[col].apply(lambda x: 6 - x if pd.notna(x) else x)

# ===============================
# 8. Create legend table
# ===============================
legend_rows = []

for var, mapping in coding_maps.items():
    for label, code in mapping.items():
        legend_rows.append({
            "Variable": var,
            "Code": code,
            "Meaning": label
        })

for label, code in likert_map.items():
    legend_rows.append({
        "Variable": "LIKERT",
        "Code": code,
        "Meaning": label
    })

legend_df = pd.DataFrame(legend_rows).drop_duplicates()

# ===============================
# 9. Save output
# ===============================
with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    df.to_excel(writer, sheet_name="Coded_Data", index=False)
    legend_df.to_excel(writer, sheet_name="Code_Legend", index=False)

print(f"Done. File saved as: {output_file}")
print("Survey columns detected:", survey_cols)
```

```python
import pandas as pd
import numpy as np

# ===============================
# 1. File names
# ===============================
input_file = "1-CSR_CMT_Raw_Data.xlsx"   # exact filename
output_file = "CSR_Coded_Data.xlsx"

# ===============================
# 2. Load data
# ===============================
df = pd.read_excel(input_file)
df.columns = df.columns.str.strip()

# ===============================
# 3. Correct coding maps
# ===============================
coding_maps = {
    "GNDR": {
        "Male": 0,
        "Female": 1
    },
    "AGE": {
        "18-24": 0,
        "25-34": 1,
        "Above 34": 2
    },
    "SIZE": {
        "Less than 50": 0,
        "50-100": 1,
        "101-200": 2,
        "201-1000": 3,
        "More than 1000": 4
    },
    "REGION": {
        "EURO": 1,
        "MENA": 2
    },
    "Degree": {
        "Undergraduate": 0,
        "Graduate": 1
    }
}

# ===============================
# 4. Correct 7-point Likert map
# ===============================
likert_map = {
    "Strongly disagree": 1,
    "Disagree": 2,
    "Somewhat disagree": 3,
    "Neutral": 4,
    "Somewhat agree": 5,
    "Agree": 6,
    "Strongly agree": 7
}

# ===============================
# 5. Copy original data
# ===============================
coded_df = df.copy()

# ===============================
# 6. Apply demographic coding
# ===============================
for col, mapping in coding_maps.items():
    if col in coded_df.columns:
        coded_df[col] = coded_df[col].astype(str).str.strip()
        coded_df[col] = coded_df[col].replace(mapping)

# ===============================
# 7. Detect survey columns
# ===============================
survey_prefixes = ["CSRSTK", "CSRCUS", "CSREMP", "EMPID", "CSRIMP", "CMTAFC"]

survey_cols = [
    col for col in coded_df.columns
    if any(col.upper().startswith(prefix) for prefix in survey_prefixes)
]

# ===============================
# 8. Apply Likert coding
# ===============================
for col in survey_cols:
    coded_df[col] = coded_df[col].astype(str).str.strip()
    coded_df[col] = coded_df[col].replace(likert_map)
    coded_df[col] = pd.to_numeric(coded_df[col], errors="coerce")

# ===============================
# 9. Reverse-code negative items
# IMPORTANT: put the actual negative items here
# For 7-point scale: reversed = 8 - value
# ===============================
negative_items = [
    # Example only — replace with your actual negative questions
    # "CSRSTK2",
    # "CSREMP4"
]

for col in negative_items:
    if col in coded_df.columns:
        coded_df[col] = coded_df[col].apply(lambda x: 8 - x if pd.notna(x) else x)

# ===============================
# 10. Build legend sheet
# ===============================
legend_rows = []

for var, mapping in coding_maps.items():
    for label, code in mapping.items():
        legend_rows.append({
            "Variable": var,
            "Type": "Demographic",
            "Original Value": label,
            "Mapped Code": code
        })

for label, code in likert_map.items():
    legend_rows.append({
        "Variable": "All survey items",
        "Type": "Likert",
        "Original Value": label,
        "Mapped Code": code
    })

legend_df = pd.DataFrame(legend_rows)

# ===============================
# 11. Build unmapped-values check
# ===============================
unmapped_rows = []

for col, mapping in coding_maps.items():
    if col in df.columns:
        original_vals = df[col].dropna().astype(str).str.strip().unique()
        for v in original_vals:
            if v not in mapping:
                unmapped_rows.append({
                    "Column": col,
                    "Unmapped Value": v
                })

for col in survey_cols:
    original_vals = df[col].dropna().astype(str).str.strip().unique()
    for v in original_vals:
        if v not in likert_map:
            unmapped_rows.append({
                "Column": col,
                "Unmapped Value": v
            })

unmapped_df = pd.DataFrame(unmapped_rows)

# ===============================
# 12. Save output
# ===============================
with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    coded_df.to_excel(writer, sheet_name="Coded_Data", index=False)
    legend_df.to_excel(writer, sheet_name="Code_Legend", index=False)
    unmapped_df.to_excel(writer, sheet_name="Unmapped_Check", index=False)

print(f"Done. File saved as: {output_file}")
print("Survey columns coded:")
print(survey_cols)

if len(unmapped_df) == 0:
    print("No unmapped values found.")
else:
    print("Some values were not mapped. Check sheet: Unmapped_Check")
```
